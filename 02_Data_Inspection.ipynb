# **DATA INSPECTION**

Data Inspection means checking the data carefully before analyzing it.
* It helps us identify missing values, duplicate records, incorrect data, and unusual values (outliers).

### **Objectives of Data Inspection**
   * check the accuracy of data.
   * Identify missing values.
   * Detect duplicate records.
   * Check data types (integer, string, float, etc.).
   * Identify incorrect or unusual values.
   * Understand the structure of the dataset.

* **Advantages of Data Inspection :**
    * Improves data quality.
    * Finds missing and duplicate values.
    * Reduces errors.
    * Saves time during data analysis.

```python
import pandas as pd

students = {
    'Name': ['Gamya Sri', 'Guna Charan', 'Gangadhar', 'Gopinath'],
    'Age': [20, 21, 20, None],
    'Marks': [85, 90, 85, 75]
}

df = pd.DataFrame(students)

print("Dataset:")
print(df)
```

```python
print("\nData Information:")
print(df.info())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

## NUMBER OF ROWS

### DEFINITION:
Number of Rows tells us how many records are available in a dataset.

* **Number of Rows is used in AI/ML for the following purposes :**

  * **Dataset Size:** To identify the total number of records available.
  * **Model Training:** To provide data for training machine learning models.
  * **Data Splitting:** To divide the dataset into training and testing data.

```python
df = pd.DataFrame(students)

print("Number of Rows:", len(df))
```

## NUMBER OF COLUMNS

### DEFINITION:
Number of Columns tells us how many features or attributes are available in a dataset.

* **Number of Columns is used in AI/ML for the following purposes:**

  * **Feature Identification:** To identify the number of features in a dataset.
  * **Model Training:** To determine how many input features are given to a machine learning model.
  * **Data Analysis:** To understand the structure of a dataset.

```python
df = pd.DataFrame(students)

print("Number of Columns:", len(df.columns))
```

## DATA TYPES

### DEFINITION:
 Data Types tell us what kind of data is stored in each column.

* **Data Types are used in AI/ML for the following purposes:**

   * **Data Understanding:** To identify the type of information stored in each column.
   * **Feature Engineering:** To transform categorical and numerical data into useful features.
   * **Error Detection:** To identify incorrect data types in a dataset.

```python
df = pd.DataFrame(students)

print("Data Types:")
print(df.dtypes)
```

## COLUMN NAMES

### DEFINITION:
 Column Names tell us what information each column contains.

 * For example, in a student dataset, Name, Age, and Marks are the column names.

* **Column Names are used in AI/ML for the following purposes:**

   * **Feature Identification:** To identify the different features in a dataset.
   * **Data Selection:** To select specific columns for analysis.
   * **Data Preprocessing:** To rename or organize columns.
   * **Model Training:** To identify input features and target columns.
   * **Data Understanding:** To understand what information is stored in the dataset.

```python
df = pd.DataFrame(students)

print("Column Names:")
print(df.columns.tolist())
```

## INDEX

### DEFINITION:
Index is used to identify the position of each row in a dataset.

* For example, in a student dataset, the rows may have index values 0, 1, 2, 3, and 4.

* **Index is used in AI/ML for the following purposes :**

  * **Row Identification:** To identify each record in a dataset.
  * **Data Selection:** To access specific rows from a dataset.
  * **Data Filtering:** To retrieve required records easily.

```python
df = pd.DataFrame(students)

print("Dataset Index:")
print(df.index)
```

## HEAD()

### DEFINITION:
head() is used to view the first few rows of a dataset.
* We can also specify the number of rows we want to display inside the parentheses.

* **The head() function is used in AI/ML for the following purposes:**
  
  * **Data Inspection:** To view the first few records of a dataset.
  * **Data Understanding:** To understand how the dataset is organized.
  * **Data Verification:** To check whether data is loaded correctly.

```python
df = pd.DataFrame(students)

print("First 2 Rows:")
print(df.head(2))
```

## TAIL()

### DEFINITION:
tail() is used to view the last few rows of a dataset.
* We can also specify the number of rows we want to display inside the parentheses.

* **The tail() function is used in AI/ML for the following purposes :**

  * **Data Inspection:** To view the last few records of a dataset.
  * **Data Understanding:** To understand how the dataset ends.
  * **Data Verification:** To check whether all records are loaded correctly.

```python
df = pd.DataFrame(students)

print("Last 2 Rows:")
print(df.tail(2))
```

## SAMPLE()

### DEFINITION:
sample() is used to select random records from a dataset.
* We can specify the number of random rows we want to display inside the parentheses.

* **The sample() function is used in AI/ML for the following purposes:**
  
   * **Data Inspection:** To view randomly selected records from a dataset.
   * **Data Understanding:** To understand different types of records.
   * **Data Sampling:** To select a smaller portion of a large dataset.

```python
df = pd.DataFrame(students)

print("Random 2 Rows:")
print(df.sample(2, random_state=1))
```

## SHAPE

### DEFINITION:
Shape tells us how many rows and columns are present in a dataset.

* **The shape attribute is used in AI/ML for the following purposes:**

   * **Data Splitting:** To verify the size of training and testing datasets.
   * **Data Preprocessing:** To check whether rows or columns have been added or removed.
