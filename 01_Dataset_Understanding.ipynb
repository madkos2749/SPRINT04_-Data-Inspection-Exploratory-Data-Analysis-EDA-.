# **DATASET UNDERSTANDING**

In simple words, Data Understanding means knowing what your data contains, what it means, and whether it is suitable for your purpose.

## DATASET

A dataset is an organized collection of related data used to understand a problem, discover patterns, perform analysis, and build machine learning models.

* Data Understanding helps us make informed decisions before modeling.

**Data quality:** Identifies missing values, duplicates, and incorrect entries.

**Feature selection:** Helps determine which columns may be useful.

**Data preparation:** Guides decisions about encoding categories, scaling numbers, and handling missing values.

* In AI/ML Dataset is used in:
   * Healthcare
   * Banking
   * E-commerce
   * Agriculture
   * Telecom

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# 1. Create a dataset
data = pd.DataFrame({
    "StudyHours": [1, 2, 3, 5, 6, 8],
    "Result": ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass"]
})

# 2. Separate input (X) and output (y)
X = data[["StudyHours"]]
y = data["Result"]

# 3. Teach the model using the dataset
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# 4. Predict for a new student
new_student = pd.DataFrame({"StudyHours": [7]})
prediction = model.predict(new_student)

print("Study hours:", 7)
print("Predicted result:", prediction[0])
```

## What is a FEATURE ?

### DEFINITION:
A Feature is an input that helps an ML model make a decision.

### Feature used in AI/ML:
  * Healthcare
  * Banking
  * E-commerce
  * Education

#### EXAMPLE:

```python
from sklearn.tree import DecisionTreeClassifier

# Feature: hours studied
X = [[1], [2], [3], [7], [8]]

# Target: exam result
y = ["Fail", "Fail", "Fail", "Pass", "Pass"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

print(model.predict([[6]])[0])
```

## What is a TARGET ?

### DEFINITION:
Target is the answer we want the model to predict.

#### EXAMPLE:


```python
from sklearn.tree import DecisionTreeClassifier

# Feature: hours studied
X = [[1], [2], [7], [8]]

# Target: exam result
y = ["Fail", "Fail", "Pass", "Pass"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

print(model.predict([[6]])[0])
```

## INDEPENDENT VARIABLES

### DEFINITION:
Independent variables are the factors we use to predict an outcome. In Machine Learning, they are commonly called features or input variables.

* In AI/ML Independent Variables are used in:
    * Banking: Transaction amount and location to detect fraud.
    * Education: Study hours and attendance to predict exam results.

#### EXAMPLE:


```python
from sklearn.linear_model import LinearRegression

# Independent variable: study hours
X = [[1], [2], [3], [4], [5]]

# Target variable: exam marks
Y = [20, 30, 40, 50, 60]

model = LinearRegression()
model.fit(X, Y)

print(model.predict([[6]])[0])
```

## Dependent Variables

### DEFINITION:
 A Dependent variable is the result we want to predict. In Machine Learning, it is commonly called the target variable or output variable.

* In AI/ML Dependent Variables used in:
    * Healthcare: Disease status predicted from patient information.
    * Banking: Fraud status predicted from transaction details.

#### EXAMPLE:

```python
from sklearn.linear_model import LinearRegression

# Independent variable: study hours
X = [[1], [2], [3], [4], [5]]

# Dependent variable: exam marks
y = [20, 30, 40, 50, 60]

model = LinearRegression()
model.fit(X, y)

print(model.predict([[6]])[0])
```

## NUMERICAL FEATURES

### DEFINITION:
Numerical Features are features expressed as numbers that an ML model can use to learn patterns and make predictions.

* In AI/ML Numerical Features used in:
    * Healthcare: Age, weight, and blood pressure to predict health risks.
    * Banking: Transaction amount to detect suspicious activity.

* **Numerical Features can be:**
    * **Discrete:** Countable values, such as the number of products purchased.
    * **Continuous:** Measurable values, such as height, weight, or temperature.

#### EXAMPLE:

```python
import pandas as pd

data = pd.DataFrame({
    "ItemsBought": [1, 3, 5],       # Discrete feature
    "ItemWeight": [1.5, 2.7, 4.2]  # Continuous feature
})

print(data)
```

## CATEGORICAL FEATURES

### DEFINITION:
Categorical Features describe what type or category something belongs to.

 * Examples include:
    * *Color:* Red, Blue, Green

    * *Payment method:* Cash, Card, UPI

    * *Department:* CSE, ECE, IT

    * *Product type:* Clothing, Electronics, Grocery

* **In AI/ML Categorical features can be used in:**

    * **Nominal:** Categories without a natural order, such as colors or cities.
    * **Ordinal:** Categories with a meaningful order, such as Low, Medium, and High.

#### EXAMPLE:

```python
import pandas as pd

data = pd.DataFrame({
    "PaymentMethod": ["UPI", "Cash", "Card", "UPI"]
})

# Convert categories into numerical columns
print("Converted INTEGER output is: ")
encoded = pd.get_dummies(data, dtype=int)
print(encoded)
print("Converted FLOAT Output is: ")
encoded = pd.get_dummies(data, dtype=float)
print(encoded)
print("Converted STRING Output is: ")
encoded = pd.get_dummies(data, dtype=str)
print(encoded)
```

## METADATA

### DEFINITION:
Metadata is information that helps us understand and manage a dataset.

* For example, a student dataset contains students' marks. Its metadata might describe the column names, data types, number of rows, and source of the data.

* **In AI/ML Metadata is used in:**
    * **Dataset management:** Records the dataset's source, size, and format.
    * **Data Understanding:** Helps identify column names, data types, and meanings.
    * **Data preprocessing:** Helps detect incorrect data types or inconsistent formats.
    * **Model development:** Records dataset versions and information needed to reproduce experiments.
    * **AI systems:** Tracks model versions, training settings, and evaluation results.

#### EXAMPLE:

```python
import pandas as pd

STORE = pd.DataFrame({
    "Product": ["Pen", "Book", "Bag"],
    "Price": [10, 50, 300]
})

# Display dataset metadata
print("Rows and columns:", STORE.shape)
print("Column names:", STORE.columns.tolist())
print("Data types:\n", STORE.dtypes)
```

## DATASET DIMENSIONS

### DEFINITION:
Dimensions tell us how many records and features a dataset contains.

* **Rows:** Represent records or observations.

* **Columns:** Represent features or variables.

For example, a dataset with 100 students and 5 columns has dimensions of (100, 5).

* **In AI/ML DATASET DIMENSIONS is used in :"**
   * **Data Understanding:** To understand the size of a dataset.
   * **Data preprocessing:** To check whether rows or columns were accidentally removed.
   * **Model training:** To understand how much data is available for learning.

#### EXAMPLE:

```python
import pandas as pd

CANDIDATES = pd.DataFrame({
    "Name": ["Rithwika", "Anshu", "Joshna"],
    "Age": [20, 21, 22],
    "Marks": [85, 90, 78]
})

print(CANDIDATES.shape)
```
