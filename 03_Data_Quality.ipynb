# DATA QUALITY ASSESSMENT

Data Quality means checking whether the data is good enough to use without errors
* It means ensuring that data is correct, complete, consistent, valid, unique, and up-to-date.

#### EXAMPLE:

```python
students = [
    {"id": 101, "name": "Ravi", "marks": 85},
    {"id": 102, "name": "Priya", "marks": 90},
    {"id": 103, "name": "", "marks": 75},
    {"id": 101, "name": "Ravi", "marks": 85},
    {"id": 104, "name": "Kumar", "marks": 120}
]

ids = set()

for s in students:
    if not s["name"]:
        print("Missing Name:", s["id"])

    elif s["id"] in ids:
        print("Duplicate ID:", s["id"])

    elif not 0 <= s["marks"] <= 100:
        print("Invalid Marks:", s["name"])

    else:
        ids.add(s["id"])
        print("Valid Record:", s["name"])
```

## MISSING VALUES

### DEFINITION:
Missing Values are the data entries that are empty, unavailable, or not recorded in a dataset.

In simple words, missing values mean some required information is not present in the data.

### EXAMPLE:

```python
import pandas as pd

persons_list = {
    "Name": ["Ranjith", "Rahul", "Preetham", "Priya Kumar", "Prudvi"],
    "Marks": [85, None, 90, 94, None]
}

df = pd.DataFrame(persons_list)

print(df.isnull())
```

## NUMBER OF MISSING VALUES

### DEFINITION:
It tells us how many data entries are missing from our dataset.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df.isnull().sum())
```

## PERCENTAGE OF MISSING VALUES

### DEFINITION:
It tells us what percentage of data is empty or unavailable.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print((df.isnull().sum() / len(df)) * 100)
```

## MISSING VALUES BY COLUMN

### DEFINITION:
It tells us which column has missing data and how many values are missing in that column.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df.isnull().sum())
```

## MISSING VALUE PATTERNS

### DEFINITION:
It helps us understand where the missing values occur and whether they follow any particular pattern.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df.isnull())
```

## DUPLICATE RECORDS

### DEFINITION:
Duplicate records mean the same data is entered multiple times in a dataset.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df.duplicated())
```

## IDENTIFY DUPLICATES

### DEFINITION:
 It helps us find out whether the same record is entered more than once in a dataset.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[df.duplicated()])
```

## COUNT DUPLICATES

### DEFINITION:
It tells us how many duplicate rows exist in a dataset.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df.duplicated().sum())
```

## IDENTIFY COMPLETELY DUPLICATE ROWS

### DEFINITION:
It helps us find records where every piece of information is repeated.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[df.duplicated(keep=False)])
```

## IDENTIFY POTENTIAL DUPLICATE RECORDS

### DEFINITION:
It helps us find records that look similar and might belong to the same person or object, even if some values are different.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[df.duplicated(subset=["Name"], keep=False)])
```

## INVALID VALUES

### DEFINITION:
Invalid values mean the data contains incorrect or unacceptable information.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[df["Marks"] > 100])
```

## NEGATIVE VALUES WHERE INAPPROPRIATE

### DEFINITION:
It helps us find values that are less than zero when negative values are not acceptable.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[df["Marks"] < 90])
```

## IMPOSSIBLE VALUES

### DEFINITION:
Impossible values mean data that is unrealistic or cannot happen in real life.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[df["Marks"] > 90])
```

## INCORRECT CATEGORIES

### DEFINITION:
Incorrect categories mean entering a value that is not included in the accepted list of values.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[~df["Name"].isin(["Ranjith", "Rahul", "Preetham", "Priya Kumar", "Prudvi"])])
```

## INVALID DATES

### DEFINITION:
Invalid dates mean dates that do not exist or are written in an incorrect format.

#### EXAMPLE:

```python
import pandas as pd

persons_list = {
    "Name": ["Ranjith", "Rahul", "Preetham", "Priya Kumar", "Prudvi"],
    "Marks": [85, None, 90, 94, None],
    "DOB" : ["2000-05-10", "31-02-2001", "2002-08-15", "2012-11-10", "20206-06-06"]
}
df = pd.DataFrame(persons_list)

df["DOB"] = pd.to_datetime(df["DOB"], errors="coerce")

print(df[df["DOB"].isnull()])
```

## INCORRECT DATATYPES

### DEFINITION:
Incorrect data types mean storing numbers as text, dates as strings, or other values in the wrong format.

#### EXAMPLE:


```python
df = pd.DataFrame(persons_list)

print(df.dtypes)

df["Marks"] = pd.to_numeric(df["Marks"])

print(df.dtypes)
```

## UNEXPECTED VALUES

### DEFINITION:
Unexpected values mean data that appears unusual or does not belong to the expected range or format.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[~df["Marks"].isin([85, None, 90, 94, None])])
```

## EMPTY STRINGS

### DEFINITION:
Empty strings mean a column contains a blank text value instead of actual information.

Example: "" represents an empty string.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[df["Name"] == ""])
```

## NULL - LIKE VALUES

### DEFINITION:
null-like values mean data that does not contain actual information, such as None, NaN, or NULL.

#### EXAMPLE:

```python
df = pd.DataFrame(persons_list)

print(df[df["Name"].isnull()])
```
